```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fruit = pd.read_table('fruit.txt')
```

```python
fruit.head()
```

```python
fruit.groupby('fruit_name').size()
```

```python
fruit.shape
```

```python
from pandas.plotting import scatter_matrix
from matplotlib import cm
features = ['mass', 'width', 'height', 'color_score']
X = fruit[features]
y = fruit['fruit_label']
cmap = plt.colormaps['gnuplot']
scatter = scatter_matrix(X , c =y , marker = 'o' , s=40 , hist_kwds = {'bins' : 15} , figsize = (9,9) , cmap = cmap)
plt.suptitle('Scatter Matrix for each input variable')
plt.savefig('scatter_matrix.png')
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
pre = model.predict(X_test)
pre
```

```python
y_test
```

```python
model.score(X_test, y_test)
```

```python
from sklearn.tree import DecisionTreeClassifier

D_model = DecisionTreeClassifier()

D_model.fit(X_train , y_train)
```

```python
D_pre = D_model.predict(X_test)
print("predict:" , D_pre)
D_model.score(X_train , y_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
K_pre = KNeighborsClassifier()
K_pre.fit(X_train,y_train)
```

```python
K_pre.score(X_test,y_test)
```
